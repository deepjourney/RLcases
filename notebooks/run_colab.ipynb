# RLcases / MARL2 — Colab GPU Runner

**使用方法：**
1. 运行时 → 更改运行时类型 → GPU
2. 按顺序执行每个 cell
3. 修改最后一格的训练参数后运行训练

```
# 1. 确认 GPU
!nvidia-smi
```

```
# 2. 挂载 Google Drive（用于保存结果）
from google.colab import drive
drive.mount('/content/drive')

DRIVE_OUTPUT = '/content/drive/MyDrive/RLcases_outputs'
!mkdir -p {DRIVE_OUTPUT}
print('结果将保存到:', DRIVE_OUTPUT)
```

```
# 3. 拉取 / 更新代码
import os

REPO_URL = 'https://github.com/deepjourney/RLcases.git'
BRANCH = 'main'
WORKDIR = '/content/RLcases'

if not os.path.exists(WORKDIR):
    !git clone --depth 1 -b {BRANCH} {REPO_URL} {WORKDIR}
else:
    %cd {WORKDIR}
    !git fetch origin {BRANCH}
    !git reset --hard origin/{BRANCH}

%cd {WORKDIR}/MARL2
!git log -1 --oneline
```

```
# 4. 安装依赖
%cd /content/RLcases/MARL2
!pip install -r requirements.txt -q
```

```
# 5. 运行训练
# 修改下面的参数来配置实验

ENV_NAME  = 'haliteM-v0'   # 环境名称
ENV_NUM   = 4              # 并行环境数量
MAX_STEPS = '1'            # 训练步数（单位：百万）
SEED      = 1

# 结果保存到 Drive
DRIVE_OUTPUT = '/content/drive/MyDrive/RLcases_outputs'

%cd /content/RLcases/MARL2
!python main.py \
    --env-name {ENV_NAME} \
    --env-num {ENV_NUM} \
    --max-stepsM {MAX_STEPS} \
    --env-seed {SEED} \
    --results-dir {DRIVE_OUTPUT} \
    --plotscore
```

```
# （可选）下次继续训练时，只需重新运行这一格拉取最新代码
%cd /content/RLcases
!git fetch origin main
!git reset --hard origin/main
!git log -1 --oneline
print('代码已更新，重新运行第 4、5 格即可继续训练')
```
