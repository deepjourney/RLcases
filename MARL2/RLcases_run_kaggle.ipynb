```python
# ===== Cell 2: 拉最新代码（每次跑实验前执行）=====
%cd /kaggle/working/RLcases
!git fetch origin main
!git reset --hard origin/main
!git log -1 --oneline
```

```python
# ===== Cell 3: 实验配置 —— 每次只改这里 =====
ENV_NAME   = 'BreakoutNoFrameskip-v4'   # 游戏名
GAMEFLAG   = 'atari'                     # atari / custom
ENV_NUM    = 16                          # 并行环境数：Kaggle 有 4 核，建议 8~16
MAX_STEPSM = 10                          # 训练步数（单位 100 万）：10 = 1000 万步
EXTRA_ARGS = '--plotscore'               # 其它开关，不需要可置空 ''

OUTPUT_DIR = '/kaggle/working/outputs'   # 输出目录（跑完在右侧 Output 面板下载）
```

```python
# ===== Cell 4: 启动训练（直接运行，不用改）=====
%cd /kaggle/working/RLcases/MARL2

cmd = (
    f'python main.py'
    f' --env-name {ENV_NAME}'
    f' --gameflag {GAMEFLAG}'
    f' --env-num {ENV_NUM}'
    f' --max-stepsM {MAX_STEPSM}'
    f' --results-dir {OUTPUT_DIR}'
    f' {EXTRA_ARGS}'
)
print('Running:', cmd)
!{cmd}
```
